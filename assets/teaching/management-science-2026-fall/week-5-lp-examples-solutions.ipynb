# Example 3-3
A company chooses skilled-worker hours, $S$, and apprentice hours, $P$, to maximize profit. Its budget must be spent exactly, skilled-worker hours cannot exceed 10,000, and apprentice hours must be at least 12,000.

$$\begin{aligned}
\max\quad & Z=0.3S+0.125P\\
\text{subject to}\quad &S+0.5P=15000,\\
&S\le10000,\quad P\ge12000,\\
&S,P\ge0.
\end{aligned}$$

**Before running the code:** Which two constraints do you expect to determine the optimal solution?

```python
# write your code here

import cvxpy as cp

# 1. Define the decision variables
S = cp.Variable()
P = cp.Variable()

# 2. Define the objective function (maximize profit)
# Maximize Z = 0.3S + 0.125P
objective = cp.Maximize(0.3 * S + 0.125 * P)

# 3. Define the constraints
constraints = [
    S + 0.5 * P == 15000,  # Budget equality
    S <= 10000,            # Upper limit on skilled-worker hours
    P >= 12000,            # Minimum apprentice hours
    S >= 0,                # Nonnegativity
    P >= 0                 # Nonnegativity
]

# 4. Define and solve the problem
problem = cp.Problem(objective, constraints)
problem.solve()

# 5. Display the results
print("Optimization status: ", problem.status)
print("Maximum profit (Z):", problem.value)
print("Optimal S:", S.value)
print("Optimal P:", P.value)
```

# Example 3-4
An architect chooses residential area, $R$, and energy-control area, $E$, to maximize total area. The model requires at least 105 square meters of residential area and 30 square meters of energy-control area, subject to a construction budget.

$$\begin{aligned}
\max\quad &R+E\\
\text{subject to}\quad &200R+400E\le24000,\\
&R\ge105,\quad E\ge30,\\
&R,E\ge0.
\end{aligned}$$

Budget and construction-cost coefficients use the same monetary unit.

**Before running the code:** Calculate the minimum budget needed to satisfy both area requirements.

```python
# write your code here

import cvxpy as cp

# 1. Define the decision variables
R = cp.Variable()
E = cp.Variable()

# 2. Define the objective function (maximize total area)
# Maximize R + E
objective = cp.Maximize(R + E)

# 3. Define the constraints
constraints = [
    200 * R + 400 * E <= 24000,  # Construction budget
    R >= 105,                     # Minimum residential area
    E >= 30,                      # Minimum energy-control area
    R >= 0,                       # Nonnegativity
    E >= 0                        # Nonnegativity
]

# 4. Define and solve the problem
problem = cp.Problem(objective, constraints)
problem.solve()

# 5. Display the results
print("Optimization status: ", problem.status)
print("Maximum total area:", problem.value)
print("Optimal R:", R.value)
print("Optimal E:", E.value)
```

# Example 3-5
A manufacturer chooses production quantities $X_1$ and $X_2$ for type A and type B motors. Each type must have production of at least 1,000 units. Type B production cannot exceed type A production by more than 500 units.

$$\begin{aligned}
\max\quad &Z=20X_1+10X_2\\
\text{subject to}\quad &X_1\ge1000,\quad X_2\ge1000,\\
&-X_1+X_2\le500,\\
&X_1,X_2\ge0.
\end{aligned}$$

**Before running the code:** Is there any constraint that limits total production?

```python
# write your code here

import cvxpy as cp

# 1. Define the decision variables
X1 = cp.Variable()
X2 = cp.Variable()

# 2. Define the objective function (maximize total profit)
# Maximize Z = 20X1 + 10X2
objective = cp.Maximize(20 * X1 + 10 * X2)

# 3. Define the constraints
constraints = [
    X1 >= 1000,                     # Minimum production of type A motors
    X2 >= 1000,                     # Minimum production of type B motors
    -X1 + X2 <= 500,               # Production-mix policy
    X1 >= 0,                        # Nonnegativity
    X2 >= 0                         # Nonnegativity
]

# 4. Define and solve the problem
problem = cp.Problem(objective, constraints)
problem.solve()

# 5. Display the results
print("Optimization status: ", problem.status)
print("Maximum total profit:", problem.value)
print("Optimal X1:", X1.value)
print("Optimal X2:", X2.value)
```

# Additional Exercise 1: A Three-Product Production Plan
A manufacturer produces three divisible products, A, B, and C. Production is measured in **tons per week**. The table gives profit and resource requirements **per ton**.

| Item | A | B | C | Weekly availability |
|---|---:|---:|---:|---:|
| Profit ($/ton) | 50 | 40 | 75 | — |
| Cutting (hours/ton) | 2 | 1 | 3 | 240 hours |
| Assembly (hours/ton) | 1 | 2 | 2 | 200 hours |
| Material (kg/ton) | 3 | 2 | 4 | 360 kg |

Contract commitments require at least 20 tons of A, 15 tons of B, and 10 tons of C. To maintain a balanced product mix, C may account for **at most 40% of total production**. There are no other demand limits. Maximize weekly profit.

## Your tasks
1. Define the decision variables, objective, and all constraints.
2. Solve the model and find the optimal solutions and value
3. Calculate the slack variables

```python
# write your code here

import cvxpy as cp

# 1. Define the decision variables (tons per week).
A = cp.Variable()
B = cp.Variable()
C = cp.Variable()

# 2. Define the objective function (weekly profit in dollars).
objective = cp.Maximize(50 * A + 40 * B + 75 * C)

# 3. Define the constraints.
constraints = [
    2 * A + B + 3 * C <= 240,
    A + 2 * B + 2 * C <= 200,
    3 * A + 2 * B + 4 * C <= 360,
    A >= 20,
    B >= 15,
    C >= 10,
    C <= 0.4 * (A + B + C),
    A >= 0, B >= 0, C >= 0
]

# 4. Define and solve the problem.
problem = cp.Problem(objective, constraints)
problem.solve()

# 5. Display the results only when a solution is available.
print("Status:", problem.status)
print(f"Production (tons): A = {A.value:.3f}, B = {B.value:.3f}, C = {C.value:.3f}")
print(f"Maximum weekly profit: ${problem.value:,.2f}")
print("Resource use and slack:")

# Cutting hours
used_cutting = 2 * A.value + B.value + 3 * C.value
available_cutting = 240
print(f"  Cutting hours: used = {used_cutting:.3f}, slack = {available_cutting - used_cutting:.3f}")

# Assembly hours
used_assembly = A.value + 2 * B.value + 2 * C.value
available_assembly = 200
print(f"  Assembly hours: used = {used_assembly:.3f}, slack = {available_assembly - used_assembly:.3f}")

# Material kg
used_material = 3 * A.value + 2 * B.value + 4 * C.value
available_material = 360
print(f"  Material kg: used = {used_material:.3f}, slack = {available_material - used_material:.3f}")

print(f"Product C share: {C.value / (A.value + B.value + C.value):.2%}")
```

# Additional Exercise 2: Transportation with Route Limits
A company ships a divisible commodity from factories A and B to customers North, Central, and East. Shipments are measured in **tons per week**. Minimize total transportation cost while meeting each customer's demand exactly. Following table describes the transportation costs from factory to customers.
| Factory | North (USD/ton) | Central (USD/ton) | East (USD/ton) | Supply capacity(tons) |
|---|---:|---:|---:|---:|
| A | 4 | 6 | 8 | 90 |
| B | 5 | 4 | 3 | 80 |
| Demand (tons) | 60 | 50 | 40 | |

Unused supply capacity is allowed. The A-to-North route can carry at most **40 tons**; the B-to-East route can carry at most **35 tons**. Other routes have no separate route limits.

## Your tasks
1. Define the decision variables, objective, and all constraints.
2. Solve the model and find the optimal solutions and value
3. Calculate the slack variables

```python
# write your code here

import cvxpy as cp

# 1. Define the six shipment variables (tons per week).
x_AN = cp.Variable()
x_AC = cp.Variable()
x_AE = cp.Variable()
x_BN = cp.Variable()
x_BC = cp.Variable()
x_BE = cp.Variable()


# 2. Define the objective function (weekly transportation cost in dollars).
transport_cost = (4*x_AN + 6*x_AC + 8*x_AE
                  + 5*x_BN + 4*x_BC + 3*x_BE)

# 3. Define supply, demand, and route-capacity constraints.
supply_and_routes = [
    x_AN + x_AC + x_AE <= 90,
    x_BN + x_BC + x_BE <= 80,
    x_AN <= 40,
    x_BE <= 35,
]
demand_constraints = [
    x_AN + x_BN == 60,
    x_AC + x_BC == 50,
    x_AE + x_BE == 40,
    x_AN >= 0, x_AC >= 0, x_AE >= 0,
    x_BN >= 0, x_BC >= 0, x_BE >= 0,
]

# 4. Define and solve the problem.
problem = cp.Problem(cp.Minimize(transport_cost),
                     supply_and_routes + demand_constraints)
problem.solve()

# 5. Display a readable shipment table.
print("Status:", problem.status)
print("Factory       North    Central       East")
print(f"A          {x_AN.value:9.3f}  {x_AC.value:9.3f}  {x_AE.value:9.3f}")
print(f"B          {x_BN.value:9.3f}  {x_BC.value:9.3f}  {x_BE.value:9.3f}")
print(f"Minimum weekly total cost: ${problem.value:,.2f}")
print(f"Unused A capacity (slack): {90-x_AN.value-x_AC.value-x_AE.value:.3f} tons")
print(f"Unused B capacity (slack): {80-x_BN.value-x_BC.value-x_BE.value:.3f} tons")
```
