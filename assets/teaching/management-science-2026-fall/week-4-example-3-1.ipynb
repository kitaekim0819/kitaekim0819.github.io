```python
# 필요한 패키지 import
import cvxpy as cp
```

```python
# 의사결정변수 정의 (Define decision variables)
X1 = cp.Variable()
X2 = cp.Variable()
```

```python
# X1 확인
X1
```

```python
# X2 확인
X2
```

```python
# 목적 함수 정의 (총 이익 극대화) (Define the objective function (Maximize total profit))
# 20 * X1 + 50 * X2 를 최대화합니다.
objective = cp.Maximize(20 * X1 + 50 * X2)
```

```python
# 제약 조건 정의 (Define constraints)
constraints = [
    1.2 * X1 + 4 * X2 <= 240,  # 조립 공정 제약식 (Assembly process constraint)
    0.5 * X1 + X2 <= 81,     # 검사 공정 제약식 (Inspection process constraint)
    X2 <= 40,                # 반도체 칩 공급 제약 (Semiconductor chip supply constraint)
    X1 >= 0,                 # 비음 제약식 (Non-negativity constraint for X1)
    X2 >= 0                   # 비음 제약식 (Non-negativity constraint for X2)
]
```

```python
# 문제 구성 (Formulate the problem)
prob = cp.Problem(objective, constraints)
```

```python
# 문제 해결 (Solve the problem) - verbose=True 옵션을 추가하여 계산 과정 출력
prob.solve() # 참고: 특별히 solver를 지정하지 않으면 CVXPY가 알아서 적절한 solver 선택함 (예시: prob.solve())
```

```python
# 중요: 최적화 결과 상태 검토. 다음 코드 실행 시 OPTIMAL이 아니라 INFEASIBLE, UNBOUNDED 등 다른 것이 나오면 코드를 잘 살펴보고 잘못된 것이 없는지 검토 필요
print("status:", prob.status)
```

```python
print("최대 이익 Z = 20X1+50X2:", prob.value)
```
