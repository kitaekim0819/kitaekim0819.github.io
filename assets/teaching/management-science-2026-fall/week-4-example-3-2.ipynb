```python
import cvxpy as cp

# 1. 의사결정 변수 정의
x1 = cp.Variable()
x2 = cp.Variable()

# 2. 목적 함수 정의 (총 식품 구입비 최소화)
z = 6000 * x1 + 1000 * x2
objective = cp.Minimize(z)

# 3. 제약 조건 정의 (최소 필요 영양소)
constraints = [
    300 * x1 + 20 * x2 >= 66,  # 단백질 최소 66g 이상
    30 * x1 + 10 * x2 >= 9,    # 철분 최소 9mg 이상
    x1 >= 0,                   # 비음 조건
    x2 >= 0,                   # 비음 조건
]

# 4. 문제 정의 및 풀이
problem = cp.Problem(objective, constraints)
problem.solve()

# 5. 결과 출력
print("최적화 상태: ",problem.status,)
print("최소 식품 구입비: ",problem.value, "원")
print("고기(X1) 최적 구매량: ", x1.value, "kg")
print("야채(X2) 최적 구매량: ", x2.value, "kg")
```
