```python
# import the cvxpy package
import cvxpy as cp

# decision variables
X1 = cp.Variable()

X2 = cp.Variable()

# objective functions
objective = cp.Maximize(20*X1+50*X2)

# constraints
constraints = [
    1.2*X1 + 4*X2 <= 240,  # assembly
    0.5*X1 + X2 <=81,  # inspection
    X2 <= 40,  # chips
    X1 >= 0,  # non-negativity
    X2 >= 0   # non-negativity
]

# Formulate the problem
prob = cp.Problem(objective, constraints)

# Solve the problem
prob.solve()

# Print the results
print("Optimal solution of X1:", X1.value)
print("Optimal solution of X2:", X2.value)
print("Optimal value:", prob.solve)
print("Optimization status:", prob.status)
```

```python
import cvxpy as cp

# decision variables
X1 = cp.Variable()
X2 = cp.Variable()

# objective functions
objective = cp.Minimize(6000*X1+1000*X2)

# constraints
constraints =[
              300*X1 + 20*X2 >= 66,
              30*X1 + 10*X2 >=9,
              X1 >= 0,
              X2 >= 0
              ]

# Formulate the problem
prob = cp.Problem(objective, constraints)

# Solve the problem
prob.solve()


# Print the results
print(prob.status, prob.value, X1.value, X2.value)
```

```python
import cvxpy as cp

S = cp.Variable()
P = cp.Variable()

objective = cp.Maximize(3000*S + 1250*P)
constraints = [
    10000*S + 5000*P == 150000000,
    S<=10000,
    P>=12000,
    S>=0,
    P>=0
]

prob = cp.Problem(objective, constraints)
prob.solve()
print(prob.status, prob.value, S.value, P.value)
```
